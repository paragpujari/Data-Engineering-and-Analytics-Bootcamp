8.  Given the dictionary:

    "id"    : 101      ,

Extract the dictionary keys into a list and the dictionary values into another list

```python
employee = {    
    "id"    : 101      ,

    "name"  : "Rahul"  ,
    
    "salary": 50000
}


##### extract all the keys from the dictionary and store it in one list
keys_ans = employee.keys()
print(keys_ans)

keys_lis = []
for i in keys_ans:
    keys_lis.append(i)
print(keys_lis)

##### extract all the values from the dictionary and store it in another list
value_ans = employee.values()
print(value_ans)

value_lis=[]
for i in value_ans:
    value_lis.append(i)

print(value_lis)
```

11.  You receive employee salary data from a CSV file as strings:

salaries = ["45000", "55000", "60000", "75000"]



Write a program to:
- Convert all salary values into integers.
- Calculate the total salary.
- Calculate the average salary.
- Identify employees earning more than ₹50,000

```python
salaries = ["45000", "55000", "60000", "75000"]

##### Convert all salary values into integers

salary_new = []
for i in salaries:
    salary_new.append(int(i))
print(salary_new)

##### Calculate the total salary
total_salary = sum(salary_new)
print(f'Total salary is {total_salary}')

##### Calculate the average salary
average_salary = total_salary/len(salary_new)
print(f'Average salary is {average_salary:.2f}')

##### Identify employees earning more than ₹50,000
res = [x for x in salary_new if(x>50000)]

print(res)
```

13.  A customer database contains inconsistent data:

Write a program to:
- Convert valid age values into integers.
- Handle invalid age values.
- Calculate the average age of customers with valid ages.

```python
customers = [
    {"name": "Amit", "age": "25"},
    {"name": "Priya", "age": "30"},
    {"name": "Ravi", "age": "Unknown"},
    {"name": "Sneha", "age": "28"}
]

##### Convert valid age values into integers
valid_ages=[]
for x in customers:
    try:
        x['age'] = int(x['age'])
        valid_ages.append(x)
    except:
        ##### Handle invalid age values
        print(f'Invalid age is {x['age']}')
print(valid_ages)

##### Calculate the average age of customers with valid ages
s=0
c = 0
for x in valid_ages:
    s = s + x['age']
    c = c + 1

print(s)
print(c)

average_ages = s/c
print(F'Average age of customers with valid ages is {average_ages:.2f}')
```

15.  A sales report contains data received from an external system:

Write a Python program to:
- Convert quantity into integers.
- Convert price into floats.
- Calculate total sales for each product.
- Calculate the overall sales amount.
- Identify the product with the highest sales.

```python
sales = [
    {"product": "Laptop", "quantity": "2", "price": "45000.50"},
    {"product": "Mobile", "quantity": "3", "price": "15000"},
    {"product": "Tablet", "quantity": "1", "price": "20000"}
]

##### Convert quantity into integers
for x in sales:
    x['quantity'] = int(x['quantity'])
    ##### Convert price into floats
    x['price']    = float(x['price'])
    ##### Calculate the total sales for each product
    x['total_sales'] = x['quantity'] * x['price']

##### Calculate the overall sales amount
total_sales_amount = 0
for x in sales:
    total_sales_amount = total_sales_amount + x['total_sales']
print(f'The total sales amount is {total_sales_amount:.2f}')

print(sales)

##### ##### Identify the product with the highest sales
max_sales_amount=0
for x in sales:
    if(x['total_sales']>max_sales_amount):
        max_sales_amount = x['total_sales']
print(sales)
print(f'Highest sales is {max_sales_amount}')
```
