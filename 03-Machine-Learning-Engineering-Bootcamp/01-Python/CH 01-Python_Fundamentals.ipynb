4.  Write a Python program to swap the values of two variables without using a third variable

```python
a=10
b=20
print('------------------------------Before Swapping-------------------------------------')
print(f'Value of a is {a}')
print(f'Value of b is {b}')
print('------------------------------After Swapping--------------------------------------')
a,b = b,a
print(f'Value of a is {a}')
print(f'Value of b is {b}')
```

5.  Write a Python program to accept a number as a string, convert it into an integer and float, and display the converted values

```python
##### enter the number as string
num = input('Enter the number')
print(num)
print(type(num))

##### Convert the number into integer and float
x_1 = int(num)
print(x_1)
print(type(x_1))

x_2 = float(num)
print(x_2)
print(type(x_2))

print(f'Original value is {num}')
print(f'------------------------------After Conversion---------------------------------------------')
print(x_1)
print(x_2)
```

7.  Write a Python program to calculate the area and perimeter of a rectangle using user-defined length and breadth

```python
##### enter the length and breadth
length = int(input('Enter the length'))
breadth = int(input('Enter the breadth'))

##### Calculate the area and perimeter of the rectangle
area = length * breadth
print(f"Area of the rectangle is {area}")

perimeter = 2*(length + breadth)
print(f"Perimeter of the rectangle is {perimeter}")
```

8.  Write a Python program to calculate the simple interest using principal, rate of interest, and time entered by the user

```python
##### enter the principal, rate and time
principal = int(input('Enter the principal'))
rate      = int(input('Enter the rate'))
time      = int(input('Enter the time'))

##### Calculate the simple interest
si = (principal*rate*time)/100.0
print(f'Simple Interest is {si}')
```

9.  An employee's monthly salary is provided as input. Write a Python program to calculate their annual salary, monthly tax deduction, and annual take-home salary based on a given tax percentage

```python
##### enter the employee monthly salary
monthly_salary = int(input('Enter the monthly salary'))
tax_percentage = float(input('Enter the tax percentage'))

##### Calculate the annual salary
annual_salary = monthly_salary*12
print(f'Annual salary of an employee : {annual_salary}')

##### Calculate the annual tax
annual_tax = (annual_salary*tax_percentage)/100.0
print(f'Annual tax of an employee : {annual_tax}')

##### Calculate the monthly tax
monthly_tax = annual_tax/12
print(f'Monthly tax of an employee : {monthly_tax}')

##### Calculate the annual take home salary
annual_take_home_salary = annual_salary - annual_tax
print(f'Annual take home salary is {annual_take_home_salary}')


##### Calculate the monthly take home salary
monthly_take_home_salary = monthly_salary - monthly_tax
print(f'Monthly take home salary is {monthly_take_home_salary}')
```

10.  An e-commerce application receives an order amount and discount percentage. Write a Python program to calculate the discount amount and final payable amount

```python
##### enter the order amount and discount percentage
order_amount = int(input('Enter the order amount'))
discount_percentage = float(input('Enter the discount percentage'))

##### Calculate the discount amount
discount_amount = (order_amount*discount_percentage)/100.0

##### Calculate the final payable amount
final_amount = order_amount - discount_amount

##### print the order details
print(f'Original order amount   :   {order_amount}')
print(f'Discount percentage     :   {discount_percentage}')
print(f'Discount_amount         :   {discount_amount}')
print(f'Final amount            :   {final_amount}')
```
